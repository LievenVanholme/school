**Recursie** is een programmeertechniek waarbij een **functie zichzelf oproept**. In plaats van een probleem op te lossen met een lus, verdeel je het probleem in kleinere deelproblemen die op dezelfde manier opgelost worden.
Een recursieve functie blijft zichzelf oproepen tot **een stopvoorwaarde** bereikt is.

Voorbeeld 1

```python
def aftellen(n):
    if n == 0:           # basisgeval
        print("Start!")
    else:
        print(n)
        aftellen(n - 1)  # recursieve stap
```

Gebruik de code:

```python
aftellen(5)
```

Voorbeeld 2

```python
def faculteit(n):
    if n == 0:
        return 1
    else:
        return n * faculteit(n - 1)
```

Gebruik van de code:

```python
print(faculteit(5))
```
