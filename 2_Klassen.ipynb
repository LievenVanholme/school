        <h2>1. Objectgeoriënteerd Programmeren (OOP) in Python</h2>

Objectgeoriënteerd programmeren (OOP) is een manier van programmeren waarbij je een programma opbouwt uit objecten. Een object stelt vaak iets voor uit de echte wereld, zoals een auto, een leerling, een bankrekening of een spelpersonage.
In plaats van alleen functies en variabelen te gebruiken, groepeer je bij OOP gegevens (data) en functionaliteit (methodes) samen in één structuur: een klasse.
Een klasse is dus een soort template voor objecten. Op basis van die ene klasse kan je dan verschillende objecten maken.

Bij python is `'str'` bijvoorbeeld een klasse. Een willekeurige zin, `z= 'zoals deze'` is dan een object van de klasse `'str'` en heeft eigenschappen die objecten van de klasse `'int'` niet hebben, zoals `z.lower()` of `z.replace('deze', 'die')`.

        <h2>3. Een klasse?</h2>
    </font>    
</div>

Een klasse is een soort template voor een verzameling van gelijkaardige objecten, bijvoorbeeld een verzameling van boeken. 
*   **Eigenschappen** (attributen) → gegevens van het object
*   **Methodes** → functies die bij het object horen

In Python maak je een klasse met het sleutelwoord class.

Algemene structuur:


```python
class NaamVanDeKlasse:

    def __init__(self, parameter1, parameter2):
        self.parameter1 = parameter1
        self.parameter2 = parameter2

    def methode(self):
        return 1   # code
```

Belangrijke onderdelen:


*   `__init__()` is de constructor, deze wordt uitgevoerd wanneer een object wordt gemaakt. Dit is een speciale python-functie, dat zie je aan de “`__`”. Deze functie wordt automatisch opgeroepen wanneer je het object aanmaakt.
*	`self` verwijst naar het huidige object
* `methode` is een eigenschap van elk object deze klasse


Nu de klasse bestaat kan je een object aanmaken en het object gebruiken. Nou ja, zoveel functionaliteit heeft het object nog niet.

```python
mijnObject = NaamVanDeKlasse(1,2)
mijnObject.methode()
```

        <h2>4. Voorbeeld van een Pythonklasse</h2>
    </font>    
</div>

We maken een klasse Leerling. Elke leerling heeft een naam en een leeftijd.  We willen ook een methode die de gegevens van de leerling toont.

```python
class Leerling:

    def __init__(self, naam, leeftijd):
        self.naam = naam
        self.leeftijd = leeftijd

    def toon_info(self):
        print("Naam:", self.naam)
        print("Leeftijd:", self.leeftijd)

    def __repr__(self):
        return f"{self.naam}"
```

Nu kunnen we een object maken van de klasse Leerling.




```python
leerling1 = Leerling("Emma", 17)
```

Wat gebeurt hier?

1.   Python maakt een nieuw object.
2.   De constructor `__init__()` wordt uitgevoerd.
3.   De waarden `"Emma"` en `17` worden opgeslagen in het object.

Nu dat leerling1 bestaat kunnen we de methode `toon_info()`oproepen.

```python
leerling1.toon_info()
```

Als we de leerling als string oproepen, wordt standaard de methode `__repr__` gebruikt.

```python
print(leerling1)
```

Een klasse kan gebruikt worden om meerdere objecten te maken.

```python
leerling1 = Leerling("Emma", 17)
leerling2 = Leerling("Lucas", 18)
```

Elke leerling heeft dus zijn eigen gegevens, maar ze gebruiken dezelfde methodes uit dezelfde klasse.

        <h2>6. Overerving</h2>
    </font>    
</div>

Een handige eigenschap van klassen is dat je ze kan gebruiken als basis voor een nieuwe klasse. Dit heet overervening.
We gebruiken de voorgaande Leerling-klasse om een steamleerling te maken

```python
class STEAMLeerling(Leerling):

    def toon_info(self):
        print( f"Naam: {self.naam} kan alles")

Busra = STEAMLeerling("Busra",17)
Busra.toon_info()
```

Pas bovenstaande code aan zodat STEAMLeerling een extra methode krijgt en pas dit toe op het Busra-object.
