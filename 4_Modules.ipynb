        <h2>1. Code hergebruiken</h2> 
    </font>    
</div>

Programmeertalen kunnen bestaande code hergebruiken.   
In C kan je een bibliotheek inladen met `#include "code.h"`
In python laad je een module in met  `import code`.

Voorbeeld:

```python
import math
print(math.pi)
```

De code zelf staat in andere bestanden ergens op de computer en noemen we bibliotheken. Door gebruik te maken van bibliotheken verhoog je de leesbaarheid van de code; de volledige computercode staat dan in verschillende blokken en bestanden in plaats van in een lang bestand.

<div>
    <font color=#690027 markdown="1">
        <h2>2. Gevaartje</h2> 
    </font>    
</div>

Als de functionaliteit van een bibliotheek verandert kan dit gevolgen hebben voor de code die gebruik maakt van de bibliotheek. Vaak loopt het programma dan beter en sneller, maar het komt ook vaak voor dat het programma dan net niet meer werkt. Dan moet je op zoek naar de oude bibliotheek die wel functioneert zoals vroeger.

<div>
    <font color=#690027 markdown="1">
        <h2>3.	Zelf een Module maken</h2> 
    </font>    
</div>

Je kan zelf een bibliotheek maken. Daarvoor schrijf je één of meerdere methodes en zet die in een aparte file. Hier bijvoorbeeld in de file `calc.py`.

```python
# calc.py
def add(x, y):
    return (x+y)

def subtract(x, y):
    return (x-y)
```

In elke andere file kan je dan deze code gebruiken:

```python
import calc
print(calc.add(10, 2))
```

Bovenstaande code werkt niet in Jupyter notebooks, want de server kan de calc.py file niet vinden. Als je de code in een andere .ipynb - file zet, lukt onderstaande wel.

```python
%run ./calc.ipynb
print(add(10, 2))
```

        <h2>4.	Plaats</h2> 
    </font>    
</div>

Python zoekt niet de hele computer af om te weten waar de module staat, maar kijkt enkel in enkele folders die in zijn systeem zitten. Deze mappen kan je zien door:

```python
import sys
for p in sys.path:
    print(p)
```
